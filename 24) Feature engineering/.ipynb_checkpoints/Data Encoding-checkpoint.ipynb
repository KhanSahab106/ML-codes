# One Hot Encoding

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df=pd.DataFrame({
    'color':['red','blue','green','green','red','blue']
})
```

```python
#create an instance of onehot encode
encoder=OneHotEncoder()
encoded=encoder.fit_transform(df[['color']]).toarray()
```

```python
encoder_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
encoder_df
```

```python
encoder.transform([['blue']]).toarray()
```

```python
pd.concat([df,encoder_df],axis=1)
```

```python
import seaborn as sns
df1=sns.load_dataset('tips')
```

```python
df1.head()
```

```python
df1['sex'].unique()
```

```python
encoded=encoder.fit_transform(df1[['sex']]).toarray()
```

```python
encode_df=pd.DataFrame(encoded,columns=encoder.get_feature_names_out())
```

```python
pd.concat([df,encode_df],axis=1)
```

# Label Coding

```python
df=pd.DataFrame({
    'color':['red','blue','green','green','red','blue']
})
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
lbl_encoder=LabelEncoder()
```

```python
lbl_encoder.fit_transform(df[['color']])
```

```python
lbl_encoder.transform([['blue']])
```

# Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df=pd.DataFrame({
   'size':['small','medium','large','medium','small','large'] 
})
```

```python
df
```

```python
encoder=OrdinalEncoder(categories=[['small','medium','large']])
```

```python
encoded=encoder.fit_transform(df[['size']])
```

# Target Guided ordinal Encoding

```python
df=pd.DataFrame({
    'city':['New york','London','Paris','Tokyo','New york','Paris'],
    'price':[200,150,300,250,180,320]
})
```

```python
df
```

```python
mean_price=df.groupby('city')['price'].mean().to_dict()
```

```python
mean_price
```

```python
df['city_encoded']=df['city'].map(mean_price)
```

```python
df
```
