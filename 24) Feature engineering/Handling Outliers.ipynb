## 5 number summary

```python
import numpy as np
list_marks=[34,56,23,67,34,78,89,45,76,34,65,45,78,35,75]
minimum,Q1,median,Q3,maximum=np.quantile(list_marks,[0,0.25,0.50,0.75,1])
```

```python
minimum,Q1,median,Q3,maximum
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
lower_fence=Q1-1.5*IQR
higher_fence=Q3+1.5*IQR
```

```python
lower_fence,higher_fence
```

```python
import seaborn as sns
sns.boxplot(list_marks)
```
