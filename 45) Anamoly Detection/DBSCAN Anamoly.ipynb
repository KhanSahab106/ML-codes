```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN
```

```python
from sklearn.datasets import make_circles
```

```python
X,y=make_circles(n_samples=750,noise=0.1,factor=0.3)
```

```python
X
```

```python
plt.scatter(X[:,0],X[:,1],c=y)
```

```python
## feature scaling
scaler=StandardScaler()
scaled_df=scaler.fit_transform(X)
scaled_df
```

```python
dbscan=DBSCAN(eps=0.2)
dbscan.fit(scaled_df)
```

```python
dbscan.labels_
```

```python
anm_idx=np.where(dbscan.labels_<0)
plt.scatter(scaled_df[:,0],scaled_df[:,1],c=dbscan.labels_)
plt.scatter(scaled_df[anm_idx,0],scaled_df[anm_idx,1],edgecolors='r')
```
