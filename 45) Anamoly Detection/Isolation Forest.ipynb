```python
import pandas as pd
df=pd.read_csv('healthcare.csv')
df.head()
```

```python
import matplotlib.pyplot as plt 
plt.scatter(df.iloc[:,0],df.iloc[:,1])
```

```python
from sklearn.ensemble import IsolationForest
isol=IsolationForest(contamination=0.2)
isol.fit(df)
predictions=isol.predict(df)
```

```python
predictions
```

```python
import numpy as np
anm_idx=np.where(predictions<0)
anm_idx 
```

```python
x=df.values
```

```python
anm_idx=np.where(predictions<0)
plt.scatter(df.iloc[:,0],df.iloc[:,1])
plt.scatter(x[anm_idx,0],x[anm_idx,1],edgecolors='r')
```
