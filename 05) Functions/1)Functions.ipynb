```python
lst=[1,2,3,4,5,6,7,8,9]
def evenoddsum(lst):
    even_sum=0
    odd_sum=0
    for i in lst:
        if i%2==0:
            even_sum+=i
        else:
            odd_sum+=i
    return even_sum,odd_sum
```

## Map() function

used to map a function to a collection(lst,tuple,set etc.)            
map format --> map(function,collection)


```python
def even_or_odd(num):
    if num%2==0:
        return "{} is even".format(num)
    else:
        return "{} is odd".format(num)
```

```python
even_or_odd(45)
```

```python
lst=[23,56,23,654,234,653,345,653,563,4536]

map(even_or_odd,lst)   #first give the function, next the iterables
```

```python
#it's a lazy function, we have to make it work to give proper output
```

```python
list(map(even_or_odd,lst))
```

```python
# mapping multiple iterables

nums1=[1,2,3]
nums2=[4,5,6]

added_nums=list(map(lambda x,y:x+y , nums1,nums2))
added_nums
```

```python
# num to string
num1=[1,2,3,4,5]
str1=list(map(str,num1))
str1
```

```python
words=['apple','banana','cherry','dragon fruit']
up_words=list(map(str.upper,words))
up_words
```

```python
def get_name(person):
    return person['name']

people=[{'name':'Abc','age':23},{'name':'Efg','age':43}]
names=list(map(get_name,people))
names
```

## Lambda Function

Function with no name
used when there is a single operation to be performed with the help of function

```python
def add(a,b):
    return a+b
```

```python
add(2,3)
```

```python
addition=lambda a,b:a+b
```

```python
addition(34,23)
```

```python
even_no=lambda a:a%2==0
even_no(6)
```

```python
add=lambda a,b,c:a+b+c
add(2,3,4)
```

```python
lst=[1,2,3,4,5]
lst_sq=list(map(lambda x:x**2,lst))
lst_sq
```

## Filter Function

used to filter elements in a collection based on a list                                                  
format --> filter(condition,collection)

```python
def even(num):
    if num%2==0:
        return True
```

```python
lst=[1,2,3,4,5,6,7,8,9]
```

```python
filter(even,lst)
```

```python
list(filter(even,lst)) # this is also a lazy function
```

```python
list(filter(lambda num:num%2==0 , lst))
```

```python
print(list(filter(lambda x:x>5 and x%2==0,lst)))
```

```python
def get_age(person):
    return person['age']>25

print(list(filter(get_age,people)))
```
