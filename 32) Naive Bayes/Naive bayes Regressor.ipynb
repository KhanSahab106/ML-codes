```python
# dependent and independent features
X=df[['total_bill', 'tip', 'sex', 'smoker', 'day','size']]
y=df['time']
X
```

```python
type(y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=10)
```

```python
X_train.head()
```

```python
## feature encoding(Label encoding and one-hot encoding)
from sklearn.preprocessing import LabelEncoder as LE

le1=LE()
le2=LE()
le3=LE()
```

```python
X_train['sex']=le1.fit_transform(X_train['sex'])
X_train['smoker']=le2.fit_transform(X_train['smoker'])
#y_train=le3.fit_transform(y_train.values)
```

```python
y_train
```

```python
X_train.head()
```

```python
X_test['sex']=le1.transform(X_test['sex'])
X_test['smoker']=le2.transform(X_test['smoker'])
#y_test=le3.transform(y_test.values)
```

```python
y_test
```

```python
## one hot encoding -- column transformer

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct=ColumnTransformer(transformers=[('Onehot',OneHotEncoder(drop='first'),[4])],remainder='passthrough')
```

```python
X_train=ct.fit_transform(X_train)
```

```python
X_test=ct.transform(X_test)
```

```python
X_train
```

```python
from sklearn.naive_bayes import GaussianNB
gnb=GaussianNB()
gnb.fit(X_train,y_train)
y_pred=gnb.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print('r2_score = ',r2_score(y_test,y_pred))
print('mean absolute error = ',mean_absolute_error(y_test,y_pred))
```
