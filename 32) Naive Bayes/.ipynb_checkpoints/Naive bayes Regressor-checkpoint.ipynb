```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## datset --> tips
df=sns.load_dataset('tips')
```

```python
# dependent and independent features
X=df.iloc[:,1:]
y=df.iloc[:,0]
X
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=10)
```

```python
X_train.head()
```

```python
## feature encoding(Label encoding and one-hot encoding)
from sklearn.preprocessing import LabelEncoder as LE

le1=LE()
le2=LE()
le3=LE()
```

```python
X_train['sex']=le1.fit_transform(X_train['sex'])
X_train['smoker']=le2.fit_transform(X_train['smoker'])
X_train['time']=le3.fit_transform(X_train['time'])
```

```python
X_train.head()
```

```python
X_test['sex']=le1.transform(X_test['sex'])
X_test['smoker']=le2.transform(X_test['smoker'])
X_test['time']=le3.transform(X_test['time'])
```

```python
## one hot encoding -- column transformer

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
ct=ColumnTransformer(transformers=[('Onehot',OneHotEncoder(drop='first'),[3])],remainder='passthrough')
```

```python
X_train=ct.fit_transform(X_train)
```

```python
X_train
```

```python
X_test=ct.transform(X_test)
```

```python
X_test
```

```python
## SVR
from sklearn.svm import SVR
svr=SVR()
```

```python
svr.fit(X_train,y_train)
```

```python
y_pred=svr.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print('r2_score = ',r2_score(y_test,y_pred))
print('mean absolute error = ',mean_absolute_error(y_test,y_pred))
```

```python
## Hyperparasmeter using Grid Search CV
from sklearn.model_selection import GridSearchCV

param_grid={'C':[0.1,1,10,100,1000],
           'gamma':[1,0.1,0.01,0.001,0.0001],
           'kernel':['rbf','linear','sigmoid']}
```

```python
grid=GridSearchCV(SVR(),param_grid=param_grid,refit=True,verbose=3)

grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid_pred=grid.predict(X_test)
```

```python
from sklearn.metrics import r2_score,mean_absolute_error
print('r2_score = ',r2_score(y_test,grid_pred))
print('mean absolute error = ',mean_absolute_error(y_test,grid_pred))
```
