# Generators

```python
def square(n):
    for i in range(3):
        yield i**2
```

```python
print(square(3))
```

```python
for i in square(3):
    print(i)
```

```python
def my_generator():
    yield 1
    yield 2
    yield 3
```

```python
gen=my_generator()
gen
```

```python
next(gen)
```

```python
for val in gen:
    print(val)
```

```python
## practical example : reading large files

def read_large_file(file_path):
    with open(file_path,'r') as file:
        for line in file:
            yield line
```

```python
file_path='large_file.txt'

for line in read_large_file(file_path):
    print(line.strip())
```

# Decorators

```python
## function Copy
## closures
## decorators
```

```python
## function copy

def welcome():
    print('welcome to the party')

welcome()
```

```python
wel=welcome
print(wel)
print(wel())
del welcome
print(wel())
wel
```

```python
## closures --> method in method with return of sub method

def main_welcome():
    msg='Welcome'
    def sub_welcome():
        print('welcome to the party')
        print(msg)
        print('enjoy the party')
    return sub_welcome()

main_welcome()
```

```python
def main_welcome(func):
    def sub_welcome():
        print('welcome to the party')
        func('hope u enjoy')
        print('enjoy the party')
    return sub_welcome()

main_welcome(print)  #this print replaces the func in main_welcome and becomes print('...')
```

```python
def main_welcome(func,lst):
    def sub_welcome():
        print('welcome to the party')
        print(func(lst))
        print('enjoy the party')
    return sub_welcome()

main_welcome(len,[1,2,3,4,5,6])
```

## Decorators

```python
def main_welcome(func):
    def sub_welcome():
        print('welcome to the party')
        func()
        print('enjoy the party')
    return sub_welcome()
```

```python
def intro():
    print('intro is given by king')

intro()
```

```python
main_welcome(intro)
```

```python
@main_welcome
def intro():  #--> tihs is going in main_welcome as parameter
    print('intro is given by king')
```

```python
## decorator

def my_decorator(func):
    def wrapper():
        print('this is wrapper before the function')
        func()
        print('this is the wrapper after the function')
    return wrapper
```

```python
@my_decorator
def say_hello():
    print('hello')
```

```python
say_hello()
```

```python
## decorator with arguments
def repeat(n):
    def decorator(func):
        def wrapper(*args,**kwargs):
            for i in range(n):
                func(*args,**kwargs)
        return wrapper

    return decorator
```

```python
@repeat(3)
def say_hello():
    print('hello')
```

```python
say_hello()
```
