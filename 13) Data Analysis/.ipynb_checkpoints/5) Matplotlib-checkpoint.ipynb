```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[1,4,9,16,25]

plt.plot(x,y)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Line Plot')
plt.show()
```

```python
# customized line plot

plt.plot(x,y,color='red',linestyle='-.',marker='*',linewidth=4,markersize=10)
plt.grid(True)
```

```python
## Multiple plots
x=[1,2,3,4,5]
y1=[1,4,9,16,25]
y2=[1,2,3,4,5]

plt.figure(figsize=(9,5))

plt.subplot(2,2,1)
plt.plot(x,y1,color='green')
plt.title("plot1")

plt.subplot(2,2,2)
plt.plot(x,y2,color='blue')
plt.title("plot2")

plt.subplot(2,2,3)
plt.plot(y1,y2,color='red')
plt.title("plot3")

plt.subplot(2,2,4)
plt.plot(y2,x)
plt.title("plot2")
```

```python
# bar plot
categories=['A','B','C','D','E']
values=[5,7,9,3,6]

plt.bar(categories,values)
```

```python
# Histogram

data=[1,2,3,4,5,3,2,4,1,3,2,4,1]

plt.hist(data,bins=5,color='orange',edgecolor='black')
```

```python
# scatter plot

plt.scatter(x,y,marker='x')
```

```python
#pie chart

labels=categories
size=[23,56,23,56,78]
colors=['red','black','blue','lightskyblue','yellowgreen']
explode=(0.1,0.2,0.9,0,0)

#creating a pychart
plt.pie(size,labels=labels,colors=colors,autopct='%1.1f%%',shadow=2,explode=explode)
```

```python
## Sales Data Visualisation

import pandas as pd
sales_data=pd.read_csv('sales_data.csv')

sales_data.head(5)
```

```python
# plot ttl sales by product

total_sales_by_product=sales_data.groupby('Product Category')['Total Revenue'].sum()
print(total_sales_by_product)
```

```python
total_sales_by_product.plot(kind='bar',color='teal')
```

```python
#plot sales trend over time

sales_trend=sales_data.groupby('Date')['Total Revenue'].sum().reset_index()
plt.plot(sales_trend['Date'],sales_trend['Total Revenue'])
```
