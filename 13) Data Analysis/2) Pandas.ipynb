```python
import pandas as pd
```

```python
## series
## one-dimensional array

data=pd.Series([1,2,3,4,5])
print(data)
print(type(data))
```

```python
## create a series from dict

data={'a':1,'b':2}
series_dict=pd.Series(data)
series_dict
```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
## DataFrame
## create a data frame from dict of lst

data={
    'Name':['abc','cde','efg'],
    'Age':[25,45,23],
    'City':['Pune','Jaipur','Mumbai']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
## create a dataframe from a lst of dict

data=[
    {'Name':'abc','age':25,'City':'Pune'},
    {'Name':'cde','age':45,'City':'Jaipur'},
    {'Name':'efg','age':23,'City':'Mumbai'}
]
df=pd.DataFrame(data)
print(df)
print(type(df))
```
