```python
import sqlite3
```

```python
connection=sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()
```

```python
# create a table
cursor.execute('''
Create Table if Not Exists Employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department text
    )
''')

## commit changes
connection.commit()
```

```python
cursor.execute('''
Select * from employees
''')
```

```python
## insert the data in sqlite table

cursor.execute('''
Insert Into employees(id,name,age,department)
values(123,'Mohd',24,'ML')
''')

cursor.execute('''
Insert Into employees(id,name,age,department)
values(124,'Abrar',24,'CS')
''')
cursor.execute('''
Insert Into employees(id,name,age,department)
values(125,'Khan',24,'DB')
''')

#commit canges
connection.commit()
```

```python
# querying th data from table
cursor.execute('select * from employees')
rows=cursor.fetchall()

#print the quried data
for row in rows:
    print(row)
```

```python
## Update the data in Tale
cursor.execute('''
Update employees
set age=22
where name='Mohd'
''')

connection.commit()
```

```python
cursor.execute('select * from employees')
rows=cursor.fetchall()

#print the quried data
for row in rows:
    print(row)
```

```python
# delete data from table
cursor.execute('''
delete from employees
where id=125
''')

connection.commit()
```

```python
cursor.execute('select * from employees')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
## Working Wwith Sales Data
# Connect to an SQLite database
connection = sqlite3.connect('sales_data.db')
cursor = connection.cursor()

# Create a table for sales data
cursor.execute('''
CREATE TABLE IF NOT EXISTS sales (
    id INTEGER PRIMARY KEY,
    date TEXT NOT NULL,
    product TEXT NOT NULL,
    sales INTEGER,
    region TEXT
)
''')

# Insert data into the sales table
sales_data = [
    ('2023-01-01', 'Product1', 100, 'North'),
    ('2023-01-02', 'Product2', 200, 'South'),
    ('2023-01-03', 'Product1', 150, 'East'),
    ('2023-01-04', 'Product3', 250, 'West'),
    ('2023-01-05', 'Product2', 300, 'North')
]

cursor.executemany('''
insert into sales(date,product,sales,region)
values(?,?,?,?)
''',sales_data)

connection.commit()
```

```python
cursor.execute('select * from sales')
rows=cursor.fetchall()

for row in rows:
    print(row)
```

```python
## close the connection
connection.close()
```
