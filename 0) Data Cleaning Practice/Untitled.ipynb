```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('train.csv',index_col='Id')
```

```python
df.head()
```

```python
df.info()
```

```python
drop_features=[feature for feature in df.columns if df[feature].isnull().mean()*100>=70]
print(drop_features)
```

```python
df.drop(drop_features,axis=1,inplace=True)
```

```python
na_features=[feature for feature in df.columns if df[feature].isnull().sum()>=1]
for feature in na_features:
    print(feature,np.round(df[feature].isnull().mean()*100,5),'% null values')
```

```python
from sklearn.linear_model import LinearRegression
regression=LinearRegression()
```

```python
known_frontage=df[df['LotFrontage'].notnull()]
missing_frontage=df[df['LotFrontage'].isnull()]
```

```python
x_train=known_frontage[['LotArea']]
y_train=known_frontage['LotFrontage']
x_test=missing_frontage[['LotArea']]
```

```python
regression.fit(x_train,y_train)
```

```python
y_pred=regression.predict(x_test)
```

```python
df.loc[df['LotFrontage'].isnull(),'LotFrontage']=y_pred
```
