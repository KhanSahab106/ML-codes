```python
import sys
def maximum_score(a,c):
    n=len(a)
    answer=sum(a)-n*c
    a.sort()
    for i in range(n//2):
        if a[i]>=c:break
        answer+=c-a[i]
    return answer
def main():
    input=sys.stdin.buffer.readline
    t=int(input())
    results=[]
    for _ in range(t):
        n,c=map(int,input().split())
        a=list(map(int,input().split()))
        results.append(str(maximum_score(a,c)))
    print("\n".join(results))
if __name__=="__main__":
    main()
```
