```python
df.drop(columns=['All Time Peak','Peak','Artist','Tour title','Ref.'],inplace=True)
```

```python
df.columns=df.columns.str.strip()
```

```python
df.columns.tolist()
```

```python
df.columns=df.columns.str.replace('\xa0',' ')
```

```python
df['Actual gross']=df['Actual gross'].replace(r'\D','',regex=True).astype(int)
```

```python
df['Adjusted gross (in 2022 dollars)']=df['Adjusted gross (in 2022 dollars)'].replace(r'\D','',regex=True).astype(int)
```

```python
df['Average gross']=df['Average gross'].replace(r'\D','',regex=True).astype(int)
```

```python
df.columns=df.columns.str.replace(r'[()]','',regex=True)
```

```python
repr(df['Years'][0])
```

```python
df.columns.tolist()
```

```python
df['Years']=df['Years'].replace(r'[^0-9]','-',regex=True)
def duration(Years):
    if '-' in Years:
        start_year=int(Years.split('-')[0])
        end_year=int(Years.split('-')[1])
        duration=end_year-start_year+1
    else:
        duration=1
        start_year=int(Years)
        end_year=int(Years)
    return pd.Series([duration,start_year,end_year])

df[['Duration','start_year','end_year']]=df['Years'].apply(duration)
```
