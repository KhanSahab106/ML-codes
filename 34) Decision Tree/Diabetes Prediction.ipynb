```python
from sklearn.datasets import load_diabetes
```

```python
dataset=load_diabetes()
```

```python
dataset
```

```python
print(dataset['DESCR'])
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
X=pd.DataFrame(dataset['data'],columns=['age','sex','bmi','bp','s1','s2','s3','s4','s5','s6'])
```

```python
X
```

```python
y=dataset['target']
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=10)
```

```python
X_train.corr()
```

```python
import seaborn as sns
plt.figure(figsize=(25,15))
sns.heatmap(X_train.corr(),annot=True)
```

```python
from sklearn.tree import DecisionTreeRegressor
regressor=DecisionTreeRegressor()
```

```python
regressor.fit(X_train,y_train)
```

```python
y_pred=regressor.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import mean_absolute_error as mae
print(mae(y_test,y_pred))
from sklearn.metrics import mean_squared_error as mse,r2_score as r2
print(mse(y_test,y_pred))
print(r2(y_test,y_pred))
```

# Hyperparameter Tuning

```python
params={
    'criterion':['squared_error', 'friedman_mse', 'absolute_error', 'poisson'],
    'splitter':['best', 'random'],
    'max_depth':[3,4,5,6],
    'max_features':['int','sqrt','log2']
}
```

```python
from sklearn.model_selection import GridSearchCV
regressor=DecisionTreeRegressor()
grid=GridSearchCV(regressor,param_grid=params,cv=5,scoring='neg_mean_squared_error')
```

```python
grid.fit(X_train,y_train)
```

```python
print(grid.best_params_)
print(grid.best_score_)
```

```python
y_predict=grid.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error as mae
print(mae(y_test,y_predict))
from sklearn.metrics import mean_squared_error as mse,r2_score as r2
print(mse(y_test,y_predict))
print(r2(y_test,y_predict))
```

```python
model=DecisionTreeRegressor(criterion='poisson', max_depth= 5, max_features= 'sqrt', splitter='random')
```

```python
model.fit(X_train,y_train)
```

```python
import matplotlib.pyplot as plt
from sklearn import tree
plt.figure(figsize=(45,30))
tree.plot_tree(model,filled=True)
```
