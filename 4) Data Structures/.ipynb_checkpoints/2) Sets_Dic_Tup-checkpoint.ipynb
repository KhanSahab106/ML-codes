## Dictionary

```python
dic={}
print(type(dic))
```

```python
dic={1:'a',2:'b',3:'c',4:'d'}
```

```python
dic
```

```python
dicti={'a':1,'b':2,'c':3,'d':4,'a':5}
print(dicti)

# using get() method
print(dicti.get('b'))
print(dicti.get('f','absent'))
```

```python
dic[1]
```

```python
for i in dic:
    print(i)
```

```python
for i in dic.values():
    print(i)
```

```python
for i in dic.items():
    print(i)
```

```python
dic[4]='e'
print(dic[4])
```

```python
del dicti['a']
print(dicti)
```

```python
#dictionary methods

keys=dicti.keys()
print(keys)
values=dicti.values()
print(values)

items=dicti.items()
print(items)
```

### Shallow copy

```python
dicti_copy=dicti
print(dicti)
print(dicti_copy)
```

```python
dicti['c']=5
print(dicti)
print(dicti_copy)
```

```python
dicti={1:'a',2:'b',3:'c',4:'d'}
dicti_copy=dicti.copy()  ## this creates a shallow copy
print(dicti_copy)
print()
dicti_copy[2]='e'
print(dicti)
print(dicti_copy)
```

### Nesteed dictionary

```python
ndic={1:{'a':'i'},2:{'b':'ii'},3:{'c':'iii'},4:{'d':'iv'}}
print(ndic)
```

```python
# another example

d1={1:'a'}
d2={2:'b'}
d3={3:'c'}

nedic={'i':d1,'ii':d2,'iii':d3}
print(nedic)
```

```python
print(nedic['ii'][2])
```

```python
# iterating over bested dictionary

for i,j in ndic.items():
    print(f"{i}:{j}")
    for k,l in j.items():
        print(f"{k}:{l}")
```

```python
# dictionary comprehension

squares={x:x**2 for x in range(10) if x%2==0}
print(squares)
```

```python
## practical examples

numbers=[1,2,2,3,3,3,4,4,4,4]
frequency={}

for number in numbers:
    if number in frequency:
        frequency[number]+=1
    else:
        frequency[number]=1

print(frequency)
```

```python
# merge dictionaries

dict1={'a':1,'b':2}
dict2={'c':3,'d':4}

merge_dict={**dict1,**dict2}
print(merge_dict)
```

## Tuples

```python
# tuples are immutable
# whole tuple can be replaced, individual values can't be changed
# indexing is possible
#tuples and list are inter-convertible
```

```python
t1=tuple()
print(type(t1))
```

```python
t2=(1,2,3,4)
print(t2)
```

```python
t2=(1,1,2,2,3,4,5,4,5,6)
print(t2.count(2))   # .count is used to find total no of occurences
```

```python
t2.index(3)
```

```python
t2
```

```python
t3=(1,4,2,3,6,4,5,8,7)
```

```python
print(t3.index(4))  # .index() finds location of element in list
print(t3[4])        #  tuple[] is used to to find element present at the location
```

```python
t3.index(7)
```

```python
lst=list(t3)
lst
```

```python
t4=tuple(lst)
t4
```

```python
# slicing works in tuples also

t4[1:4:2]
```

```python
# operations

concatenated_tuple=t2+t3
concatenated_tuple
```

```python
print(t4*3)
```

```python
# packing a tuple

packed_tuple=1,'Hello',3.14
print(packed_tuple)
```

```python
#unpacking a tuple

a,b,c=packed_tuple
print(a)
print(b)
print(c)

#this is the unpacking of tuple
```

```python
# unpacking with *

numbers=(1,2,3,4,5,6)
first,*middle,last=numbers
print(first)
print(middle)
print(last)
```

```python
#nested Tuple,   nested list also exist

tpl=((1,2,3),('a','b','c'),(True,False))

print(tpl[1])
print(tpl[2][1])
```

```python
lst=[[1,2,3,4,5],['a','b','c'],['A','B','C'],(True,False)]
print(lst[2][2])
print(lst[3][1])
```

```python
#iterating over nested tuple

for i in tpl:
    for j in i:
        print(i,j)
```
