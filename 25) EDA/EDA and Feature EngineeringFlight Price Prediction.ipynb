```python
## Feature Engineering
df['Date']=df['Date_of_Journey'].str.split('/').str[0]
df['Month']=df['Date_of_Journey'].str.split('/').str[1]
df['Year']=df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.head()
```

```python
df.info()
```

```python
df['Date']=df['Date'].astype(int)
df['Month']=df['Month'].astype(int)
df['Year']=df['Year'].astype(int)
```

```python
df.info()
```

```python
## drop date of DOJ

df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df.head()
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(" ")[0])
```

```python
df['Arrival_Hour']=df['Arrival_Time'].str.split(':').str[0].astype(int)
df['Arrival_Minute']=df['Arrival_Time'].str.split(':').str[1].astype(int)
df['Dep_Hour']=df['Dep_Time'].str.split(':').str[0].astype(int)
df['Dep_Minute']=df['Dep_Time'].str.split(':').str[1].astype(int)
```

```python
df.info()
```

```python
df.drop(['Arrival_Time','Dep_Time'],axis=1,inplace=True)
```

```python
df.info()
```

```python
df['Total_Stops'].unique()
```

```python
df['Total_Stops']=df['Total_Stops'].map({'non-stop':0,'1 stop':1,'2 stops':2,'3 stops':3,'4 stops':4,np.nan:1})
```

```python
df[df['Total_Stops'].isnull()]
```

```python
df.drop('Route',axis=1,inplace=True)
```

```python
def extract_hours(duration):
    if 'h' in duration:
        return int(duration.split(' ')[0].split('h')[0])
    else:
        return 0

df['duration_hours']=df['Duration'].apply(extract_hours)
```

```python
def extract_minutes(duration):
    if 'm' in duration:
        if 'h' in duration:
            return int(duration.split(' ')[1].split('m')[0])
        else:
            return int(duration.split('m')[0])
    else:
        return 0

df['duration_minutes']=df['Duration'].apply(extract_minutes)
```

```python
df[df['duration_hours'].isnull()]
```

```python
df[df['duration_minutes'].isnull()]
```

```python
df['duration']=(60*df['duration_hours'])+(df['duration_minutes'])
```

```python
df.head()
```

```python
df.drop(['Duration','duration_hours','duration_minutes'],axis=1,inplace=True)
```
