### Assignment 3: Querying Data

1. Write a Python function to fetch and display all records from the `employees` table.

```python
def fetch_all():
    conn=sqlite3.connect('test.db')
    cursor=conn.cursor()
    cursor.execute('select * from employees')
    rows=cursor.fetchall()
    for row in rows:
        print(row)
    conn.close()
```

2. Write a Python function to fetch and display all employees from a specific department.

```python
cursor.execute('''select * from employees
where department='HR'
''')
rows=cursor.fetchall()

for row in rows:
    print(row)

connection.close()
```

### Assignment 4: Updating Data

1. Write a Python function to update the department of an employee based on their `id`.

```python
def updating_data(age,id):
    conn=sqlite3.connect('test.db')
    cursor=conn.cursor()
    cursor.execute('''
    update employees
    set age=?
    where id=?
    ''',(age,id))
    conn.commit()
    conn.close()

updating_data(35,123)
```

### Assignment 5: Deleting Data

1. Write a Python function to delete an employee from the `employees` table based on their `id`.

```python
def delete_record(id):
    conn=sqlite3.connect('test.db')
    cursor=conn.cursor()
    cursor.execute('''
    delete from employees
    where id=?''',(id,)
    )
    conn.commit()
    conn.close()
```

### Assignment 6: Advanced Queries

1. Write a Python function to fetch and display employees older than a certain age.

```python
def fetch_greater_than_age(age):
    conn=sqlite3.connect('test.db')
    cursor=conn.cursor()
    cursor.execute('''
    select * from employees
    where age>?''',(age,)
    )
    rows=cursor.fetchall()
    for row in rows:
        print(row)
    conn.close()

fetch_greater_than_age(27)
```

2. Write a Python function to fetch and display employees whose names start with a specific letter.

```python
def fetch_starting_with_letter(letter):
    conn=sqlite3.connect('test.db')
    cursor=conn.cursor()
    cursor.execute('''
    select * from employees
    where name like ?
    ''',(letter+'%',))
    rows=cursor.fetchall()
    for row in rows:
        print(row)
    conn.close()

fetch_starting_with_letter('m')
```

### Assignment 7: Handling Transactions

1. Write a Python function to insert multiple employees into the `employees` table in a single transaction. Ensure that if any insertion fails, none of the insertions are committed.

```python
def insert_multiple(employees):
    conn=sqlite3.connect('test.db')
    cursor=conn.cursor()
    try:
        cursor.executemany('''
        insert into employees(id,name,age,department)
        values(?,?,?,?)
        ''',employees)
        conn.commit()
        print('successfully inserted')
    except Exception as e:
        conn.rollback()
        print('error occoured, transaction rolled back')
        print(e)
    finally:
        conn.close()

employees = [
    (6, 'Frank', 40, 'Finance'),
    (7, 'Grace', 29, 'Engineering'),
    (8, 'Hannah', 35, 'Marketing'),
    (9, 'Ivan', 38, 'Sales'),
    (6, 'Jack', 45, 'HR') 
]
insert_multiple(employees)
```

```python
fetch_all()
```

2. Write a Python function to update the age of multiple employees in a single transaction. Ensure that if any update fails, none of the updates are committed.

```python
def updating_multiple_age(updates):
    conn=sqlite3.connect('test.db')
    cursor=conn.cursor()
    cursor.executemany('''
    update employees
    set age=?
    where id=?
    ''',updates)
    conn.commit()
    conn.close()

# test
updates=[
    (42,6),
    (37,7)
]

updating_multiple_age(updates)
```

```python
cursor.execute('''
create table employees(
id integer,
name text,
age integer,
department text,
department_id integer,
foreign key(department_id) references departments(id)
)
''')

cursor.execute('''
insert into employees (id,name,age,department)
select id,name,age,department from old_employees
''')
conn.commit()
```
