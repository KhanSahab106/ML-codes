### Assignment 4: Data Aggregation

1. Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

```python
df4=pd.DataFrame({'Category':np.random.choice(['A','B','C'],size=10),'Values':np.random.randint(1,11,size=10)})
print('original DataFrame:\n',df4)

sum=df4.groupby('Category').sum()
print(sum)
mean=df4.groupby('Category').mean()
print(mean)
```

### Assignment 5: Merging DataFrames

1. Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

```python
df5=pd.DataFrame({'key':[1,2,3,4,5],'Value1':[6,7,8,9,10],'Value2':[11,12,13,14,15]})
print(df5)
df_5=pd.DataFrame({'key':[1,2,3,4,5],'Value3':[16,17,18,19,20]})
print(df_5)

merged=pd.merge(df5,df_5,on='key')
print(merged)
```

### Assignment 6: Time Series Analysis

1. Create a Pandas DataFrame with a datetime index and one column filled with random integers. Resample the DataFrame to compute the monthly mean of the values.

```python
date_rng=pd.date_range(start='2025-01-01',end='2025-12-31',freq='D')
df6=pd.DataFrame(date_rng,columns=['date'])
df6['data']=np.random.randint(1,101,size=(len(date_rng)))
display(df6.head(10))

df6.set_index('date',inplace=True)
display(df6.head(5))

Monthly_Mean=df6.resample('M').mean()
Monthly_Mean
```

### Assignment 8: Pivot Tables

1. Create a Pandas DataFrame with columns 'Date', 'Category', and 'Value'. Create a pivot table to compute the sum of 'Value' for each 'Category' by 'Date'.

```python
date_rng=pd.date_range(start='2025-01-01',end='2025-12-31',freq='D')
df8=pd.DataFrame({'date':np.random.choice(date_rng,size=20),
                 'Category':np.random.choice(['A','B','C'],size=20),
                'Value':np.random.randint(1,1000,size=20)})

display(df8)

df8.pivot_table(values='Value',index='date',columns='Category',aggfunc='sum',fill_value=0)
```

2. Create a Pandas DataFrame with columns 'Year', 'Quarter', and 'Revenue'. Create a pivot table to compute the mean 'Revenue' for each 'Quarter' by 'Year'.

```python
date_rng=pd.date_range(start='1905',end='2025',freq='YE')
df82=pd.DataFrame({'Year':np.random.choice(date_rng,size=20),
                   'Quater':np.random.choice(['Q1','Q2','Q3','Q4'],size=20),
                   'Revenue':np.random.randint(1,1000,size=20)})
display(df82)

df82.pivot_table(values='Revenue',index='Year',columns='Quater',aggfunc='sum',fill_value=0)
```
