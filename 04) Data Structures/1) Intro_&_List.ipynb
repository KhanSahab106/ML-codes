# List Comprehension

### Traditional Method

```python
lst=[1,2,3,4,5,6,7]
lst1=[]
def lst_square(lst):
    for i in lst:
        lst1.append(i*i)
    return lst1
```

```python
lst_square(lst)
```

## Using List Comprehension

in square braces enter the value to return followed by a fore loop and zore or more for or if clause.

```python
list2=[i*i for i in lst]
print(list2)
```

```python
list3=[i*i for i in lst if i%2==0]
```

```python
list3
```

```python
#### Nested list comprehension

lst1=[1,2,3,4]
lst2=['a','b','c','d']

pair=[(i,j) for i in lst1 for j in lst2]
print(pair)
```

```python
lst3=['a','bc','def','ghij','klmno','pqrstu','vwxyz']
lengths=[len(i) for i in lst3]
print(lengths)
```
