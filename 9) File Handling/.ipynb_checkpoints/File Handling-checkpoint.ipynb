```python
# reading a file

with open('example.txt','r') as file:
    content=file.read()
    print(content)
```

```python
with open('example.txt','r') as file:
    for line in file:
        print(line)
```

```python
with open('example.txt','r') as file:
    for line in file:
        print(line.strip())
```

```python
#writing a file(over writing)

with open('example.txt','w') as file:
    file.write('hellow world \n')
    file.write('this is a new line')
```

```python
with open('example.txt','r') as file:
    for line in file:
        print(line.strip())
```

```python
# without overwriting

with open('example.txt','a') as file:
    file.write('\nappend operation taking place')
```

```python
with open('example.txt','r') as file:
    for line in file:
        print(line.strip())
```

```python
# writing a list of files
lines=['\n first line','\n second line','\n third line']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
with open('example.txt','r') as file:
    for line in file:
        print(line.strip())
```

```python
## Binary file

data=b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
with open('example.bin','rb') as file:
    content=file.read()
    print(content)
```

```python
## read the content from source and write to destination

with open('example.txt','r') as source_file:
    content=source_file.read()

with open('destination.txt','w') as destination_file:
    destination_file.write(content)
    
```

```python
with open('destination.txt','r') as file:
    content=file.read()
    print(content)
```

```python
with open('destination.txt','r') as file:
    lines=file.readlines()
    line_count=len(lines)
    word_count=sum(len(i.split()) for i in lines)
    char_count=sum(len(i) for i in lines)
    print('line count: ',line_count,'\n word count: ',word_count,'\n char count: ',char_count)
```

```python
path='example1.txt'
if os.path.exists(path):
    print(f"the path '{path}' exists")
else:
    print(f"the path {path} doesn't exist")
    
```

```python
# Checking if path is a file or directory
import os
path='example.txt'
if os.path.isfile(path):
    print(f"the path {path} is file ")
elif os.path.isdir(path):
    print('the path',path,'is directory')
else:
    print('the path ',path ,' is neither a file nor a directory')
```

```python
## getting absolute path
relative_path='example.txt'
absolute_path=os.path.abspath(relative_path)
print(absolute_path)
```
